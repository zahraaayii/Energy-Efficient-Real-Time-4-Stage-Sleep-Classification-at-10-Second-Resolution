```python
selected_features=[ 'HRV_Prc20NN_x',  'HRV_SDANN1_y',
       'HRV_CVSD_y', 'HRV_MCVNN_y', 'HRV_SDRMSSD_y', 'HRV_Prc80NN_y',
       'HRV_pNN20_y', 'HRV_MinNN_y', 'HRV_MaxNN_y', 'HRV_CSI_Modified_y',
       'HRV_IALS_y', 'HRV_PSS_y', 'HRV_PI_y', 'HRV_DFA_alpha1_y',
       'HRV_ShanEn_y', 'HRV_FuzzyEn_y', 'HRV_CD_y', 'HRV_HFD_y', 
       'var_y', 'rmse_y', 'skewness_y', 
       'impulse_factor_y', 'margin_factor_y', 'rms_y','peak_to_peak_y','mean_y','peak_factor_y','median_y','max_y','waveform_factor_y',
       'HRV_MeanNN_y','HRV_MeanNN_x','HRV_MedianNN_y',
       'HRV_VLF_y',
       'HRV_ApEn_y', 
       'anns']
```

```python
for file in data_base:
    i=pd.read_csv(base_dir+f'feature/{file}_win.csv')
    i= i[selected_features]
    i = i.replace((np.inf, -np.inf, np.nan), 0).reset_index(drop=True)
    i.to_csv(base_dir+f'feature/{file}_win_sel.csv', index=False)
i
```

```python
from sklearn.model_selection import LeaveOneGroupOut
from sklearn.metrics import accuracy_score, f1_score, cohen_kappa_score

X_list=[]
y_list=[]
subjects=[]

for sid,file in enumerate(data_base):

    df=pd.read_csv(base_dir+f'feature/{file}_win_sel.csv')
    df = df.replace((np.inf, -np.inf, np.nan), 0).reset_index(drop=True)
    X=df.drop(columns=['anns'])
    y=df['anns']

    X_list.append(X.values)
    y_list.append(y.values)

    subjects.extend([sid]*len(df))

X=np.vstack(X_list)
y=np.hstack(y_list)
subjects=np.array(subjects)

print("Total samples:",X.shape)
```

```python
logo=LeaveOneGroupOut()

accs=[]
f1s=[]
kappas=[]

for train_idx,test_idx in logo.split(X,y,groups=subjects):

    X_train,X_test=X[train_idx],X[test_idx]
    y_train,y_test=y[train_idx],y[test_idx]
    
    
    print("Train subjects:",np.unique(subjects[train_idx]))
    print("Test subject:",np.unique(subjects[test_idx]))
    
    std_slc = StandardScaler()
    model1=RandomForestClassifier(n_estimators=50,n_jobs=-1)
    model = Pipeline(steps=[('std_slc', std_slc),
                           ('m', model1)])
    model.fit(X_train,y_train)

    y_pred=model.predict(X_test)
    
    print(classification_report(y_test,y_pred))

    print("-----") 
    accs.append(accuracy_score(y_test,y_pred))
    f1s.append(f1_score(y_test,y_pred,average='macro'))
    kappas.append(cohen_kappa_score(y_test,y_pred))

print("LOSO Accuracy:",np.mean(accs))
print("LOSO Macro F1:",np.mean(f1s))
print("LOSO Kappa:",np.mean(kappas))
```
